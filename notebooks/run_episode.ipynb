```python
import sys
sys.path.insert(0, "src/kaggriculture")

from kaggle_environments import make
from agent import agent

env = make("kaggriculture", configuration={"episodeSteps": 720}, debug=True)
steps = env.run([agent, "random"])

final_state = steps[-1][0]
print("Episode finished.")
print("Final reward:", final_state.get("reward"))
print("If this is still exactly 3000.0, restart the session (see the note above) and re-run from the top.")
```

```python
import json

first_obs = steps[0][0]["observation"]
with open("sample_observation.json", "w") as f:
    json.dump(first_obs, f, indent=2, default=str)
print("Wrote sample_observation.json (first turn) for reference")
```
