```python
import sys
sys.path.insert(0, "src/kaggriculture")

from kaggle_environments import make
from agent import agent

# VERIFY: the exact registered environment id -- "kaggriculture" is a guess
# based on the competition's name. If make() raises, check
# kaggle_environments.envs or the competition page for the real id.
env = make("kaggriculture")
steps = env.run([agent, "random"])

final_state = steps[-1][0]
print("Episode finished.")
print("Final reward (should reflect farm['money'] at episode end):", final_state.get("reward"))
```
